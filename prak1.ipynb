```python
#soal keempat
def hitung_lingkaran(jari_jari):
    
    
    pi = 3.14159
    
    luas = pi * jari_jari ** 2
    keliling = 2 * pi * jari_jari
    
    return {
        'luas': luas,
        'keliling': keliling
    }

# Contoh penggunaan
jari_jari = 10
hasil = hitung_lingkaran(jari_jari)

print(f"Jari-jari: {jari_jari}")
print(f"Luas lingkaran: {hasil['luas']:.2f}")
print(f"Keliling lingkaran: {hasil['keliling']:.2f}")
```

```python
#soal kelima
# Meminta input usia dari pengguna
usia = int(input("Masukkan usia: "))

# Mengecek kategori usia
if usia >= 0 and usia <= 13:
    print("Kategori: Anak")
elif usia >= 14 and usia <= 24:
    print("Kategori: Remaja")
elif usia >= 25 and usia <= 49:
    print("Kategori: Dewasa")
elif usia >= 50:
    print("Kategori: Lansia")
else:
    print("Usia tidak valid!")
    
```
